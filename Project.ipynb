```python
#Part 1
##first implement cylic shift function
def cyclic_shift(ciphertext,shift):
    plaintext=""
    for i in ciphertext:
        plaintext=plaintext+alphabet[(alphabet.index(i)+shift)%27]
    return plaintext
##apply function to text 1, shift of 9 was manually found
print(cyclic_shift(ciphertexts[1],9))
```

```python
#Part 2
##import text to personal server
import re
with open('moby.txt', 'r', encoding='utf-8', errors='ignore') as f:
    moby = f.read()
##simplify text to only uppercase and spaces
moby = moby.upper()
sample_text = re.sub('[^A-Z ]', ' ', moby)
sample_text = re.sub(' +', ' ', sample_text)
##print test on beginning of text
print(len(sample_text))
print(sample_text[:500])
print(set(sample_text))
```

```python
#Part 3
##count frequencies of characters in altered Moby Dick text
from collections import Counter
english_counts = Counter(sample_text)
english_ranked = [ch for ch, count in english_counts.most_common()]
print(english_ranked)
```

```python
##count frequencies of characters from cipher text
msg = ciphertexts[2]   
msg_counts = Counter(msg)
def frequency_decode(msg, english_ranked):
    msg_ranked = [ch for ch, count in msg_counts.most_common()]
    ##compare lists and interchange
    mapping = {}
    for k in range(len(msg_ranked)):
        mapping[msg_ranked[k]] = english_ranked[k]
    decoded = ''.join(mapping[ch] for ch in msg)
    print(decoded)
```

This mapping didn't completly solve the cipher, however with a few switched characters it is almost there, the most common charcaters should be correct, but those with similar frequencies may need some switching around.

```python
#Part 4 begun
def swap(mapping, x, y):
    """Swap the roles characters x and y in the mapping."""
    new_map = dict(mapping)
    for c, p in mapping.items():
        if p == x:
            new_map[c] = y
        elif p == y:
            new_map[c] = x
    return new_map
```

```python
# Part 5
# count all consecutive bigrams in cleaned moby dick text
bigram_counts = Counter(zip(sample_text, sample_text[1:]))

# put bigram frequencies into a 27x27 matrix and single character counts into a vector
bigram_matrix = np.array([[bigram_counts[(i,j)] for j in alphabet] for i in alphabet])
char_freq = np.array([english_counts[i] for i in alphabet])

# calculate p(i,j) = (freq(ij) + 1) / freq(i) and take the natural log
# dividing by char_freq[:, None] divides each row i by freq(i)
p_matrix = (bigram_matrix + 1) / char_freq[:, None]
log_p = np.log(p_matrix)

# store log probabilities in a dictionary using lookup label (i, j) for quick lookup in score()
log_p_dict = {
    (a, b): float(log_p[i, j])
    for i, a in enumerate(alphabet)
    for j, b in enumerate(alphabet)
}

# computethe log-likelihood score S(m) for any string m
def score(msg):
    return sum(log_p_dict[pair] for pair in zip(msg, msg[1:]))
```

The decoded message recieves a much higher score than the raw cyphertext, showing that the bigram model can distinguish real English text apart from cyphertext. Computing the log probabilities in advance means that `score()` runs quickly.
