```python
#Part 1
##first implement cylic shift function
def cyclic_shift(ciphertext,shift):
    plaintext=""
    for i in ciphertext:
        plaintext=plaintext+alphabet[(alphabet.index(i)+shift)%27]
    return plaintext
##apply function to text 1, shift of 9 was manually found
print(cyclic_shift(ciphertexts[1],9))
```

```python
##count frequencies of characters from cipher text
msg = ciphertexts[2]   

def frequency_decode(msg, english_ranked):
    msg_counts = Counter(msg)
    msg_ranked = [ch for ch, count in msg_counts.most_common()]
    ##compare lists and interchange
    mapping = {}
    for k in range(len(msg_ranked)):
        mapping[msg_ranked[k]] = english_ranked[k]
    decoded = ''.join(mapping[ch] for ch in msg)
    return decoded, mapping
    
##print map of swapped characters by rank    
msg = ciphertexts[2]
decoded, mapping = frequency_decode(msg, english_ranked)  
print(decoded)
```

```python
#Part 4 
##establish character swapping
def swap(mapping, x, y):
    new_map = dict(mapping)
    for c, p in mapping.items():
        if p == x:
            new_map[c] = y
        elif p == y:
            new_map[c] = x
    return new_map
    
##apply swapping to cipher text    
def decode(msg, mapping):
    return ''.join(mapping.get(ch, ch) for ch in msg)

    msg = ciphertexts[2]
decoded, mapping = frequency_decode(msg, english_ranked)
print(decoded)
##apply each swap then evaluate next move
mapping = swap(mapping, 'I', 'S')
mapping = swap(mapping, 'K', 'V')
mapping = swap(mapping, 'I', 'H')
mapping = swap(mapping, 'N', 'I')
mapping = swap(mapping, 'G', 'P')
mapping = swap(mapping, 'U', 'F')
mapping = swap(mapping, 'B', 'Y')
mapping = swap(mapping, 'W', 'F')
mapping = swap(mapping, 'Q', 'X')
print(decode(msg, mapping))
```

```python
# Part 5
# count all consecutive bigrams in cleaned moby dick text
bigram_counts = Counter(zip(sample_text, sample_text[1:]))

# put bigram frequencies into a 27x27 matrix and single character counts into a vector
bigram_matrix = np.array([[bigram_counts[(i,j)] for j in alphabet] for i in alphabet])
char_freq = np.array([english_counts[i] for i in alphabet])

# calculate p(i,j) = (freq(ij) + 1) / freq(i) and take the natural log
# dividing by char_freq[:, None] divides each row i by freq(i)
p_matrix = (bigram_matrix + 1) / char_freq[:, None]
log_p = np.log(p_matrix)

# store log probabilities in a dictionary using lookup label (i, j) for quick lookup in score()
log_p_dict = {
    (a, b): float(log_p[i, j])
    for i, a in enumerate(alphabet)
    for j, b in enumerate(alphabet)
}

# computethe log-likelihood score S(m) for any string m
def score(msg):
    return sum(log_p_dict[pair] for pair in zip(msg, msg[1:]))
```

The decoded message recieves a much higher score than the raw cyphertext, showing that the bigram model can distinguish real English text apart from cyphertext. Computing the log probabilities in advance means that `score()` runs quickly.

```python
# Looking at part 6 - consider ciphertext 8
import random
from math import *
m8 = ciphertexts[8]
# find the score of ciphertext 6 before any decryption has taken place
sm8 = score(m8)
print(sm8)

# start my part

def metropolis(message, N, print_current_guess, temperature, seed=None):
    # use a seed if required
    if seed != None:
        random.seed(seed)
    
    # use the frequency_decode function to partially decode the message, and create a mapping
    decoded0, mapping0 = frequency_decode(message, english_ranked)
    
    # add some initial values
    current_mapping = mapping0
    current_message = decoded0
    current_score = score(current_message)

    for i in range(N):
        # generate two random letters to swap
        randomx, randomy = random.randrange(0, 27, 1), random.randrange(0, 27, 1)
        x = alphabet[randomx]
        y = alphabet[randomy]
        # swap 
        intermediate1 = swap(current_mapping, x, y)
        
        intermediate2 = decode(message, intermediate1)
        # find the score of the message after the swap has occurred

        # print the message and its score every few iterations:
        new_score = score(intermediate2)
        if (i + 1) % print_current_guess == 0:
            print(current_message)
            print(current_score)
        
        # update the values if the new score is greater than the original one
        if new_score > current_score:
            current_mapping = intermediate1
            current_message = intermediate2
            current_score = new_score
        else:
            # if the new score is the same or lower than the original, then keep the change with the required
            # probability
            temperatureprobability = exp((new_score - current_score)/temperature)
            if random.random() < temperatureprobability:
                current_mapping = intermediate1
                current_message = intermediate2
                current_score = new_score
    return current_message, current_score, current_mapping

# try the algorithm with different values of T
metropolis(m8, 100000, 100000, 10, 59898)
metropolis(m8, 100000, 100000, 1, 58)
metropolis(m8, 100000, 100000, 0.1, 58)

# as we can see, the smaller value of T gives the best results as it has the highest score. Now try even smaller 
# values of T with different seeds
```

```python
# Simulated Annealing
# now define a new version of the function where we can start with T large, e.g. T=10, and end with something small
# such as T = 0.1, by reducing it by a constant ratio at each step.

def metropolis_adjustable_temperature(message, N, print_current_guess, temperature, seed=None):
    # use a seed if required
    if seed != None:
        random.seed(seed)
    
    # use the frequency_decode function to partially decode the message and create a mapping
    decoded0, mapping0 = frequency_decode(message, english_ranked)
    
    # add some initial values
    current_mapping = mapping0
    current_message = decoded0
    current_score = score(current_message)

    for i in range(N):
        # generate two random letters to swap
        randomx, randomy = random.randrange(0, 27, 1), random.randrange(0, 27, 1)
        x = alphabet[randomx]
        y = alphabet[randomy]
        # swap 
        intermediate1 = swap(current_mapping, x, y)
        
        intermediate2 = decode(message, intermediate1)
        # find the score of the message after the swap has occurred

        # print the temperature, the message and its score every few iterations:
        new_score = score(intermediate2)
        if (i + 1) % print_current_guess == 0:
            print(current_message)
            print(current_score)
            print(temperature)
        
        # update the values if the new score is greater than the original one
        if new_score > current_score:
            current_mapping = intermediate1
            current_message = intermediate2
            current_score = new_score
        else:
            # if the new score is the same or lower than the original, then keep the change with the required
            # probability
            # as long as the temperature is greater than 0.1, decrease it by 10% with each iteration:
            if temperature > 0.1:
                temperature = 0.9 * temperature
            temperatureprobability = exp((new_score - current_score)/temperature)
            if random.random() < temperatureprobability:
                current_mapping = intermediate1
                current_message = intermediate2
                current_score = new_score
    return current_message, current_score, current_mapping
```

```python
print(metropolis_adjustable_temperature(m8, 100000, 50000, 1, 1970))
print(metropolis_adjustable_temperature(m8, 100000, 50000, 10, 1970))
print(metropolis_adjustable_temperature(m8, 1000000, 500000, 10, 190))
# the results appear to be similar.
```
