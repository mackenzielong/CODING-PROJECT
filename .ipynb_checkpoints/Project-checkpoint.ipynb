```python
#Part 1
#first implement cylic shift function
def cyclic_shift(ciphertext,shift):
    plaintext=""
    for i in ciphertext:
        plaintext=plaintext+alphabet[(alphabet.index(i)+shift)%27]
    return plaintext
#apply function to text 1, shift of 9 was manually found
print(cyclic_shift(ciphertexts[1],9))
```
